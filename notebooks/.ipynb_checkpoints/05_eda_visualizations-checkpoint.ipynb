# Phase 6: Exploratory Data Analysis (EDA) and Visualizations

## Step 1: Load Cleaned Datasets

```python
import pandas as pd
import matplotlib.pyplot as plt

telco = pd.read_csv(
    "../data/processed/clean_telco.csv"
)

retail_churn = pd.read_csv(
    "../data/processed/clean_retail_churn.csv"
)

retail_sales = pd.read_csv(
    "../data/processed/clean_retail_sales.csv"
)
```

## Step 2: Create Visualizations Folder

```python
from pathlib import Path

Path(
    "../reports/telco_visualizations"
).mkdir(parents=True, exist_ok=True)

Path(
    "../reports/retail_churn_visualizations"
).mkdir(parents=True, exist_ok=True)

Path(
    "../reports/retail_sales_visualizations"
).mkdir(parents=True, exist_ok=True)
```

## TELCO EDA

### Step 3: Churn Distribution

```python
# What percentage of customers churn?
churn_counts = (
    telco["Churn Label"]
    .value_counts()
)

plt.figure(figsize=(6,4))

churn_counts.plot(
    kind="bar",
    color=["steelblue", "tomato"]
)

plt.title(
    "Telco Churn Distribution"
)

plt.ylabel("Customers")

plt.tight_layout()

plt.savefig(
    "../reports/telco_visualizations/churn_distribution.png"
)

plt.show()
```

### Step 4: Contract Type vs Churn

```python
# Which contracts experience the highest churn?
contract_churn = pd.crosstab(
    telco["Contract"],
    telco["Churn Label"]
)

contract_churn.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title(
    "Contract Type vs Churn"
)

plt.tight_layout()

plt.savefig(
    "../reports/telco_visualizations/contract_vs_churn.png"
)

plt.show()
```

### Step 5: Monthly Charges Distribution

```python
plt.figure(figsize=(8,5))

telco["Monthly Charges"].hist(
    bins=30
)

plt.title(
    "Monthly Charges Distribution"
)

plt.xlabel(
    "Monthly Charges"
)

plt.ylabel(
    "Frequency"
)

plt.tight_layout()

plt.savefig(
    "../reports/telco_visualizations/monthly_charges_distribution.png"
)

plt.show()
```

### Step 6: Tenure vs Churn

```python
telco.boxplot(
    column="Tenure Months",
    by="Churn Label",
    figsize=(8,5)
)

plt.title(
    "Tenure Months by Churn Status"
)

plt.suptitle("")

plt.savefig(
    "../reports/telco_visualizations/tenure_vs_churn.png"
)

plt.show()
```

```python
### Step 7: CLTV Distribution
```

```python
plt.figure(figsize=(8,5))

telco["CLTV"].hist(
    bins=30
)

plt.title(
    "Customer Lifetime Value Distribution"
)

plt.tight_layout()

plt.savefig(
    "../reports/telco_visualizations/cltv_distribution.png"
)

plt.show()
```

## Cross-Dataset Comparison

### Step 18: Compare Churn Rates

```python
comparison = pd.DataFrame(
    {
        "Dataset": [
            "Telco",
            "Retail Churn",
            "Retail Sales"
        ],
        "Churn Rate": [
            (
                telco[
                    "Churn Label"
                ]
                .eq("Yes")
                .mean()
            ) * 100,
            (
                retail_churn[
                    "Churn"
                ]
                .eq(1)
                .mean()
            ) * 100,
            (
                retail_sales[
                    "churned"
                ]
                .eq(1)
                .mean()
            ) * 100
        ]
    }
)

comparison.plot(
    x="Dataset",
    y="Churn Rate",
    kind="bar",
    legend=False,
    figsize=(8,5)
)

plt.title(
    "Churn Rate Comparison"
)

plt.ylabel(
    "Percent"
)

plt.tight_layout()

plt.savefig(
    "../reports/churn_rate_comparison.png"
)

plt.show()
```

## Step 20: Correlation Heatmap - Telco

```python
# Which Telco variables are most strongly associated with churn?

churn_correlations = (
    telco[
        [
            "Tenure Months",
            "Monthly Charges",
            "Total Charges",
            "CLTV",
            "Churn Value"
        ]
    ]
    .corr(numeric_only=True)
)

print(
    churn_correlations["Churn Value"]
    .sort_values(ascending=False)
)
```

```python
telco_corr_cols = [
    "Tenure Months",
    "Monthly Charges",
    "Total Charges",
    "Churn Value",
    "CLTV"
]

telco_corr = (
    telco[telco_corr_cols]
    .corr(numeric_only=True)
)

plt.figure(figsize=(8,6))

plt.imshow(
    telco_corr,
    cmap="coolwarm",
    aspect="auto"
)

plt.colorbar()

plt.xticks(
    range(len(telco_corr.columns)),
    telco_corr.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(telco_corr.columns)),
    telco_corr.columns
)

plt.title(
    "Telco Correlation Heatmap"
)

plt.tight_layout()

plt.savefig(
    "../reports/telco_visualizations/telco_correlation_heatmap.png"
)

plt.show()
```

## Step 21: Correlation Heatmap - Retail Churn

```python
retail_churn_cols = [
    "Tenure",
    "WarehouseToHome",
    "HourSpendOnApp",
    "NumberOfDeviceRegistered",
    "SatisfactionScore",
    "NumberOfAddress",
    "OrderAmountHikeFromlastYear",
    "CouponUsed",
    "OrderCount",
    "DaySinceLastOrder",
    "CashbackAmount",
    "Churn"
]

retail_corr = (
    retail_churn[
        retail_churn_cols
    ]
    .corr(numeric_only=True)
)

plt.figure(figsize=(10,8))

plt.imshow(
    retail_corr,
    cmap="coolwarm",
    aspect="auto"
)

plt.colorbar()

plt.xticks(
    range(len(retail_corr.columns)),
    retail_corr.columns,
    rotation=90
)

plt.yticks(
    range(len(retail_corr.columns)),
    retail_corr.columns
)

plt.title(
    "Retail Churn Correlation Heatmap"
)

plt.tight_layout()

plt.savefig(
    "../reports/retail_churn_visualizations/retail_churn_correlation_heatmap.png"
)

plt.show()
```

## Step 22: Correlation Heatmap - Retail Sales

```python
retail_sales["churned_numeric"] = (
    retail_sales["churned"]
    .map({
        "Yes": 1,
        "No": 0
    })
)

retail_sales_cols = [
    "age",
    "membership_years",
    "quantity",
    "unit_price",
    "avg_purchase_value",
    "online_purchases",
    "in_store_purchases",
    "total_returned_value",
    "total_sales",
    "total_transactions",
    "customer_support_calls",
    "website_visits",
    "days_since_last_purchase",
    "churned_numeric"
]

retail_sales_corr = (
    retail_sales[
        retail_sales_cols
    ]
    .corr(numeric_only=True)
)

plt.figure(figsize=(12,10))

plt.imshow(
    retail_sales_corr,
    cmap="coolwarm",
    aspect="auto"
)

plt.colorbar()

plt.xticks(
    range(len(retail_sales_corr.columns)),
    retail_sales_corr.columns,
    rotation=90
)

plt.yticks(
    range(len(retail_sales_corr.columns)),
    retail_sales_corr.columns
)

plt.title(
    "Retail Sales Correlation Heatmap"
)

plt.tight_layout()

plt.savefig(
    "../reports/retail_sales_visualizations/retail_sales_correlation_heatmap.png"
)

plt.show()
```

## Step 24: Generate EDA Summary Report

### Step 1: Capture Key Metrics

#### Telco

```python
telco_churn_rate = (
    telco["Churn Value"].mean()
) * 100

telco_corr = (
    telco[
        [
            "Tenure Months",
            "Monthly Charges",
            "Total Charges",
            "CLTV",
            "Churn Value"
        ]
    ]
    .corr()
)

telco_top_corr = (
    telco_corr["Churn Value"]
    .drop("Churn Value")
    .abs()
    .sort_values(ascending=False)
)
```

#### Retail Churn

```python
retail_churn_rate = (
    retail_churn["Churn"].mean()
) * 100

retail_top_corr = (
    retail_corr["Churn"]
    .drop("Churn")
    .abs()
    .sort_values(ascending=False)
)
```

#### Retail Sales

```python
retail_sales_churn_rate = (
    retail_sales["churned_numeric"]
    .mean()
) * 100

retail_sales_top_corr = (
    retail_sales_corr["churned_numeric"]
    .drop("churned_numeric")
    .abs()
    .sort_values(ascending=False)
)
```

### Step 2: Generate Report from Results

```python
from pathlib import Path

report = Path(
    "../reports/eda_summary.md"
)

with open(
    report,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "# Exploratory Data Analysis Summary\n\n"
    )

    # ===================================
    # TELCO
    # ===================================

    f.write(
        "## Telco Customer Churn\n\n"
    )

    f.write(
        f"- Churn Rate: {telco_churn_rate:.2f}%\n"
    )

    f.write(
        f"- Average Monthly Charges: "
        f"${telco['Monthly Charges'].mean():.2f}\n"
    )

    f.write(
        f"- Average CLTV: "
        f"${telco['CLTV'].mean():.2f}\n\n"
    )

    f.write(
        "### Strongest Churn Correlations\n\n"
    )

    for feature, corr in (
        telco_top_corr.head(5)
        .items()
    ):

        f.write(
            f"- {feature}: "
            f"{corr:.3f}\n"
        )

    # ===================================
    # RETAIL CHURN
    # ===================================

    f.write(
        "\n## Retail Customer Churn\n\n"
    )

    f.write(
        f"- Churn Rate: {retail_churn_rate:.2f}%\n"
    )

    f.write(
        f"- Average Cashback: "
        f"{retail_churn['CashbackAmount'].mean():.2f}\n"
    )

    f.write(
        f"- Average Satisfaction Score: "
        f"{retail_churn['SatisfactionScore'].mean():.2f}\n\n"
    )

    f.write(
        "### Strongest Churn Correlations\n\n"
    )

    for feature, corr in (
        retail_top_corr.head(5)
        .items()
    ):

        f.write(
            f"- {feature}: "
            f"{corr:.3f}\n"
        )

    # ===================================
    # RETAIL SALES
    # ===================================

    f.write(
        "\n## Retail Sales\n\n"
    )

    f.write(
        f"- Churn Rate: "
        f"{retail_sales_churn_rate:.2f}%\n"
    )

    f.write(
        f"- Average Total Sales: "
        f"${retail_sales['total_sales'].mean():.2f}\n"
    )

    f.write(
        f"- Average Transactions: "
        f"{retail_sales['total_transactions'].mean():.2f}\n\n"
    )

    f.write(
        "### Strongest Churn Correlations\n\n"
    )

    for feature, corr in (
        retail_sales_top_corr.head(5)
        .items()
    ):

        f.write(
            f"- {feature}: "
            f"{corr:.3f}\n"
        )

    # ===================================
    # CROSS DATASET
    # ===================================

    f.write(
        "\n## Cross-Industry Comparison\n\n"
    )

    f.write(
        f"- Telco Churn Rate: "
        f"{telco_churn_rate:.2f}%\n"
    )

    f.write(
        f"- Retail Churn Rate: "
        f"{retail_churn_rate:.2f}%\n"
    )

    f.write(
        f"- Retail Sales Churn Rate: "
        f"{retail_sales_churn_rate:.2f}%\n"
    )

    f.write(
        "\n### Key Finding\n\n"
    )

    f.write(
        "The strongest churn predictors identified through correlation analysis "
        "will be used during Phase 7 predictive modeling.\n"
    )

print(
    f"Created: {report}"
)
```
