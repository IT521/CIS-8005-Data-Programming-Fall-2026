# Phase 9: Final Report Compilation and Executive Dashboard Preparation

## Customer Churn Analytics Project

### Step 1: Imports and Setup

```python
from pathlib import Path
import pandas as pd

final_dir = Path(
    "../reports/final"
)

final_dir.mkdir(
    parents=True,
    exist_ok=True
)

print(
    f"Created: {final_dir}"
)
```

### Step 2: Helper Function

```python
def load_text_file(path):

    with open(
        path,
        "r",
        encoding="utf-8"
    ) as f:

        return f.read()
```

### Step 3: Load Previous Phase Reports

```python
sqlite_report = load_text_file(
    "../reports/sqlite_analysis_summary.md"
)

mongodb_report = load_text_file(
    "../reports/mongodb_analysis_summary.md"
)

comparison_report = load_text_file(
    "../reports/sql_vs_mongodb_comparison.md"
)

eda_report = load_text_file(
    "../reports/eda_summary.md"
)

predictive_report = load_text_file(
    "../reports/predictive_modeling_summary.md"
)

feature_report = load_text_file(
    "../reports/feature_importance_summary.md"
)

recommendation_report = load_text_file(
    "../reports/business_recommendations.md"
)
```

#### Load Model Results

```python
# Load Phase 7 model results

model_results = pd.read_csv(
    "../reports/model_comparison.csv"
)

# Best model per dataset based on F1

best_models_by_dataset = (
    model_results
    .sort_values(
        "F1",
        ascending=False
    )
    .groupby(
        "Dataset"
    )
    .first()
    .reset_index()
)

# Best overall model

best_overall_model = (
    model_results
    .sort_values(
        "F1",
        ascending=False
    )
    .iloc[0]
)

display(
    best_models_by_dataset
)
```

### Step 4: Create Final Project Report

```python
final_report = final_dir / "final_project_report.md"

with open(
    final_report,
    "w",
    encoding="utf-8"
) as f:

    f.write(
"""# Customer Churn Analytics Project

## Executive Summary

This project analyzed customer churn across three industries:

- Telecommunications
- Retail Customer Churn
- Retail Sales

The project combined:

- Data Cleaning and Feature Engineering
- SQLite Analytics
- MongoDB Analytics
- Exploratory Data Analysis
- Machine Learning
- Feature Importance Analysis
- Business Intelligence Recommendations

Key findings include:

- Telco Churn Rate: 26.54%
- Retail Customer Churn Rate: 16.84%
- Retail Sales Churn Rate: 49.97%

The strongest churn drivers included:

- Monthly Charges
- Customer Tenure
- Cashback Incentives
- Customer Value

Decision Tree models achieved the strongest predictive performance, while Random Forest models provided feature importance analysis used to generate business recommendations.

---

"""
    )

    sections = [
        ("SQLite Analysis", sqlite_report),
        ("MongoDB Analysis", mongodb_report),
        ("SQL vs MongoDB Comparison", comparison_report),
        ("Exploratory Data Analysis", eda_report),
        ("Predictive Modeling", predictive_report),
        ("Feature Importance Analysis", feature_report),
        ("Business Recommendations", recommendation_report)
    ]

    for title, content in sections:

        f.write(
            f"\n# {title}\n\n"
        )

        f.write(content)

        f.write(
            "\n\n---\n\n"
        )

    f.write(
    """
    # Project Conclusions
    
    ## Key Outcomes
    
    - Decision Tree produced the strongest churn prediction performance.
    - Monthly Charges was the strongest Telco churn driver.
    - Tenure was the strongest Retail Customer Churn driver.
    - Retail Sales requires additional feature engineering.
    
    ## Future Work
    
    - Evaluate XGBoost
    - Explore K-Means segmentation
    - Investigate additional Retail Sales features
    - Deploy dashboard-driven monitoring
    """
    )

print(
    f"Created: {final_report}"
)
```

### Step 5: Executive Dashboard Summary

```python
dashboard_file = (
    final_dir /
    "executive_dashboard_summary.md"
)

with open(
    dashboard_file,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "# Executive Dashboard Summary\n\n"
    )

    # ===================================
    # CHURN RATES
    # ===================================

    f.write(
        "## Churn Rates\n\n"
    )

    f.write(
"""| Dataset | Churn Rate |
|----------|----------:|
| Telco | 26.54% |
| Retail Churn | 16.84% |
| Retail Sales | 49.97% |

"""
    )

    # ===================================
    # BEST MODEL TABLE
    # ===================================

    f.write(
        "## Best Performing Models\n\n"
    )

    f.write(
        best_models_by_dataset[
            [
                "Dataset",
                "Model",
                "F1"
            ]
        ]
        .rename(
            columns={
                "Model": "Best Model",
                "F1": "F1 Score"
            }
        )
        .round(4)
        .to_markdown(
            index=False
        )
    )

    f.write("\n\n")

    # ===================================
    # TOP DRIVERS
    # ===================================

    f.write(
"""## Top Churn Drivers

### Telco

- Monthly Charges
- Total Charges
- CLTV

### Retail Customer Churn

- Tenure
- CashbackAmount
- WarehouseToHome

### Retail Sales

- total_sales
- unit_price
- avg_purchase_value

"""
    )

    # ===================================
    # BEST OVERALL MODEL
    # ===================================

    f.write(
        "## Best Overall Model\n\n"
    )

    f.write(
        f"- Dataset: {best_overall_model['Dataset']}\n"
    )

    f.write(
        f"- Model: {best_overall_model['Model']}\n"
    )

    f.write(
        f"- F1 Score: {best_overall_model['F1']:.4f}\n\n"
    )

    # ===================================
    # EXECUTIVE RECOMMENDATIONS
    # ===================================

    f.write(
"""## Executive Recommendations

- Review pricing and customer value strategies.
- Expand customer reward and cashback programs.
- Focus retention efforts on recently acquired customers.
- Deploy predictive churn monitoring.
- Prioritize high-risk customers for intervention.

"""
    )

    f.write(
        "## Key Project Finding\n\n"
    )

    f.write(
        "Decision Tree achieved the strongest balance of precision and recall across the analyzed churn datasets, while Random Forest models identified the primary business drivers of customer churn.\n"
    )

print(
    f"Created: {dashboard_file}"
)
```

### Step 6: Project Metrics Summary

```python
metrics_file = (
    final_dir /
    "project_metrics_summary.md"
)

telco = pd.read_csv(
    "../data/processed/clean_telco.csv"
)

retail_churn = pd.read_csv(
    "../data/processed/clean_retail_churn.csv"
)

retail_sales = pd.read_csv(
    "../data/processed/clean_retail_sales.csv"
)

telco_churn_rate = (
    telco["Churn Value"]
    .mean() * 100
)

retail_churn_rate = (
    retail_churn["Churn"]
    .mean() * 100
)

retail_sales_churn_rate = (
    retail_sales["churned"]
    .eq("Yes")
    .mean() * 100
)

metrics_df = pd.DataFrame(
    [
        ["Telco Customers", f"{len(telco):,}"],
        ["Retail Churn Customers", f"{len(retail_churn):,}"],
        ["Retail Sales Customers", f"{len(retail_sales):,}"],
        [
            "Telco Churn Rate",
            f"{telco_churn_rate:.2f}%"
        ],
        [
            "Retail Churn Rate",
            f"{retail_churn_rate:.2f}%"
        ],
        [
            "Retail Sales Churn Rate",
            f"{retail_sales_churn_rate:.2f}%"
        ],
        [
            "Best Overall Model",
            best_overall_model["Model"]
        ],
        [
            "Best F1 Score",
            f"{best_overall_model['F1']:.4f}"
        ]
    ],
    columns=[
        "Metric",
        "Value"
    ]
)

with open(
    metrics_file,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "# Project Metrics Summary\n\n"
    )

    f.write(
        metrics_df.to_markdown(
            index=False
        )
    )

print(
    f"Created: {metrics_file}"
)
```
